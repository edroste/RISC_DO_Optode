```python
def check_optode_vars(output_fdir, output_fname): 
    
    ds = xr.open_dataset(output_fdir + output_fname, decode_times=True)

    print(f"{erddap_fname}: ")
    optode_var_list = [  "BPHASE_DOXY", # Aanderaa 3830
                         "RPHASE_DOXY", # Aanderaa 3830
                         "C1RPHASE_DOXY", # Aanderaa 4330
                         "C2RPHASE_DOXY",# Aanderaa 4330
                         "TEMP_VOLTAGE_DOXY", 
                         'TEMP_DOXY',
                         'TPHASE_DOXY', # Aanderaa 4330 | nomenclature in Argo
                         "TCPHASE_DOXY",  # Aanderaa 4330
                         'DPHASE_DOXY',
                         "CALPHASE_DOXY", 
                         'MOLAR_DOXY',
                         'OXYSAT_DOXY',
                         "BPHASE_DOXY_QC", # Aanderaa 3830
                         "RPHASE_DOXY_QC", # Aanderaa 3830
                         "C1RPHASE_DOXY_QC", # Aanderaa 4330
                         "C2RPHASE_DOXY_QC",# Aanderaa 4330
                         'TEMP_DOXY_QC',
                         'TPHASE_DOXY_QC', # Aanderaa 4330 | nomenclature in Argo
                         "TCPHASE_DOXY_QC", # Aanderaa 4330
                         'DPHASE_DOXY_QC',
                         "CALPHASE_DOXY_QC", 
                         'MOLAR_DOXY_QC',
                         'OXYSAT_DOXY_QC', 
                         ]
    
    print("Optode variables with only NaNs (True) or with data (False): ")
    for var_name in optode_var_list:
        if var_name in ds.data_vars: 
            if ds[var_name].isnull().all().item(): 
                print(var_name, " exists but has no data. ")
            else: 
                print(var_name, " exists and has data. ")
        else:
            print(var_name, " not found in variable list. ")
```
